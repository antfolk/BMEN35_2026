## Numbers and calculations

Let's start with a couple of numbers. Python can add, subtract, multiply and divide them. Parentheses work in the same way as in ordinary arithmetic.

```python
print(5 + 7)
print(3.5 + 8)
print(3 * 4)
print(3 / 4)
print((3 - 4) * 7)
```

## Comments and variables

A comment starts with `#`. Python ignores it when it runs the code. Comments are useful when we want to explain why a line is there. Your future self will thank you for writing them.

We use `=` to give a value a name. Python works out the type from the value, so we do not need to write `int` or `float` first.

```python
# These are comments. They explain the code to the person reading it.
my_age = 35
my_name = 'Lund University'

print(my_age)
print(my_name)
print(type(my_age))
print(type(my_name))
```

## A small mistake to look out for

Let's store a number as text and multiply it by 12. Before running the cell, guess what the result will be. Is Python doing arithmetic here, or repeating text?

```python
monthly_income = '1000'  # The quotation marks make this a string.
yearly_income = 12 * monthly_income
print(yearly_income)
print(type(monthly_income))
```

Python did not calculate twelve thousand. It repeated the string twelve times. That is because `'1000'` is text, not a number. If we want to do arithmetic, we should store a number without quotation marks.

```python
monthly_income = 1000
yearly_income = 12 * monthly_income
print(yearly_income)
print(type(monthly_income))
```

## Lists

A list stores several values in a particular order. We write a list using square brackets. The first item is at index `0`, the second item at index `1`, and so on. Let's look at the list before selecting anything from it.

```python
cities = ['Lund', 'Stockholm', 'Malmö', 'Malmö']
print(cities)
print('First city:', cities[0])
print('Third city:', cities[2])
print('Number of items:', len(cities))
```

We can add items to a list. `append` puts one item at the end. `insert` puts an item at a chosen index. Run the cell one line at a time and watch how the list changes.

```python
numbers = [1, 3, 4]
print(numbers)

numbers.insert(1, 2)  # Put 2 at index 1.
print(numbers)

numbers.append(5)  # Add 5 at the end.
print(numbers)
```

## Tuples: an ordered collection we do not change

A tuple is similar to a list: it keeps values in order and we can select an item using its index. The difference is that we normally do not change the values after we create the tuple. Tuples use parentheses.


```python
coordinates = (55.7, 13.2)
print('Coordinates:', coordinates)
print('First value:', coordinates[0])
print('Number of values:', len(coordinates))
```

## Dictionaries

A dictionary stores values using `key: value` pairs. This can make the data easier to read than a list when each value has a name. We use the key to get a value back.

```python
patient = {
    'age': 64,
    'heart_rate': 72,
    'site': 'clinic A'
}

print(patient)
print('Age:', patient['age'])
print('Heart rate:', patient['heart_rate'])
```

## A `while` loop

A `while` loop repeats as long as its condition is true. We need to change the counter inside the loop, or it may never stop. Here we count from 1 to 4.


```python
count = 1
while count < 5:
    print('Count:', count)
    count = count + 1  # Increase the counter so the loop will eventually stop.
```

## Flow control: a `for` loop

A `for` loop repeats the indented lines once for each item in a sequence. Notice the colon after the `for` statement and the indentation on the next line. Indentation tells Python which lines belong to the loop.

```python
measurements = [72, 75, 71, 78]

for measurement in measurements:
    print('Measurement:', measurement)
```

Sometimes we also want the position of an item. `enumerate` gives us both the position and the value. Remember that the first position is zero.

```python
for position, measurement in enumerate(measurements):
    print('Position', position, 'has value', measurement)
```

## Flow control: `if`, `elif` and `else`

An `if` statement lets the code choose between paths. Python checks the conditions from top to bottom and runs the first matching block. The indentation is important here too.

```python
heart_rate = 72

if heart_rate > 100:
    print('Above this example threshold')
elif heart_rate < 60:
    print('Below this example threshold')
else:
    print('Between the example thresholds')
```

These numbers are only here to show how an `if` statement works. They are not medical thresholds. In a real health setting, a threshold needs clinical context and careful validation.

## Functions

A function is a named block of code. We define it with `def`, and it only runs when we call it. A function can take inputs (arguments) and send a result back with `return`.

```python
def add_numbers(first_number, second_number):
    # Return the sum so the caller can use it later.
    result = first_number + second_number
    return result

answer = add_numbers(5, 4)
print(answer)
```

We can also write a function for a calculation we might want to use again. This one calculates the mean of a non-empty list. Think about what would happen if we gave it an empty list.

```python
def calculate_mean(values):
    total = sum(values)
    number_of_values = len(values)
    mean = total / number_of_values
    return mean

measurements = [72, 75, 71, 78]
print('Mean measurement:', calculate_mean(measurements))
```

## Importing a library

Python comes with many libraries. A library contains code that other people have written and tested. We import the parts we need. Here we use the `math` library to calculate a square root.

```python
import math

number = 25
square_root = math.sqrt(number)
print(square_root)
```
