# Matplotlib and plotting

In this notebook we will use Matplotlib to make plots. A plot is not just something to decorate a report with. When we work with data, we should look at the data and check that it makes sense before we fit a model.

We will make a line plot, change its labels and limits, look at histograms, and make a scatter plot where color shows the class. The examples use generated data, so we know how they were made.

## Import Matplotlib

We will use `pyplot`, usually imported as `plt`. This is the common way you will see Matplotlib imported in Python examples.

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Make a first plot

Let's make a plot from a short list of values. Matplotlib puts the values in order on the horizontal axis and draws a line between them. The axes are scaled automatically.

```python
values = [0, 1, 0, 1, 0, 1, -1, 0, -1, 1]
plt.plot(values)
plt.show()
```

We can change the axes ourselves. The four numbers below are `[xmin, xmax, ymin, ymax]`. Try to connect those limits with what you see in the plot.

```python
plt.plot(values)
plt.axis([-2, 11, -2, 2])
plt.show()
```

## Plot a mathematical function

A line plot needs x-values and y-values. We can use `np.linspace` to make evenly spaced x-values, then use NumPy's exponential function to calculate y for each x.

```python
x = np.linspace(0, 5, 500)
y = np.exp(x)

plt.plot(x, y)
plt.show()
```

This plot shows the values, but it would not be clear in a report. Let's add a title, axis labels, a legend and a grid. We can also save a figure as a PNG image.

```python
plt.plot(x, y, color='red', label='exponential function')
plt.title('The exponential function')
plt.xlabel('x (arbitrary units)')
plt.ylabel('exp(x) (arbitrary units)')
plt.grid(True)
plt.legend()
plt.savefig('exponential_function.png', transparent=True)
plt.show()
```

The saved image appears in the Colab file browser. When we make a figure for other people to read, we should label what is being shown and use units when we know them.

## Histograms

A histogram groups numerical values into bins and shows how many values fall in each bin. Let's start with numbers drawn from a uniform distribution between 0 and 1.

```python
uniform_values = np.random.rand(10000)
plt.hist(uniform_values, bins=30, edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Count')
plt.title('Values from a uniform distribution')
plt.show()
```

The values are spread fairly evenly between zero and one. Now let's draw numbers from a normal distribution with a mean near zero. The shape of the histogram should look different.

```python
normal_values = np.random.randn(10000)
plt.hist(normal_values, bins=40, edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Count')
plt.title('Values from a standard normal distribution')
plt.show()
```

## Scatter plots and classes

A scatter plot shows two numerical features at once. Here we will generate four groups of points. The color of each point shows which group it belongs to.

```python
from sklearn.datasets import make_blobs

# make_blobs gives us two feature values and one group label per point.
X, y = make_blobs(n_samples=300, centers=4, n_features=2, random_state=0)
print('Feature array shape:', X.shape)
print('Label array shape:', y.shape)
print('Class labels:', np.unique(y))
```

```python
colors = ['red', 'green', 'blue', 'cyan']

for class_number in np.unique(y):
    points = X[y == class_number]
    plt.scatter(
        points[:, 0], points[:, 1],
        color=colors[class_number],
        label=f'Class {class_number}',
        edgecolor='black', alpha=0.7
    )

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Generated data: color shows the class')
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

The colors look quite separate here because we asked `make_blobs` to create separated groups. Real data are often messier. When classes overlap, a plot can help us see that no simple line will separate everything perfectly.

## Two plots in one figure

Sometimes we want to compare two signals. `subplots` gives us separate axes in one figure. We use the same time axis for both plots so their timing can be compared.

```python
time = np.linspace(0, 2, 500)
slow_wave = np.sin(2 * np.pi * time)
combined_wave = slow_wave + 0.25 * np.sin(2 * np.pi * 5 * time)

figure, axes = plt.subplots(2, 1, figsize=(9, 5), sharex=True)
axes[0].plot(time, slow_wave)
axes[0].set_ylabel('Amplitude')
axes[0].set_title('Slow wave')
axes[1].plot(time, combined_wave, color='darkorange')
axes[1].set_xlabel('Time (seconds)')
axes[1].set_ylabel('Amplitude')
axes[1].set_title('Two waves added together')
plt.tight_layout()
plt.show()
```
