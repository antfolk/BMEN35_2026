This exercise illustrates an important idea: individual observations can have a distribution that looks different from the distribution of sample means. With repeated samples of the same size, the sample means tend to cluster around the population mean. As the sample size grows, the sample means usually vary less. These patterns are related to the central limit theorem, but the plots here are an illustration rather than a proof.
